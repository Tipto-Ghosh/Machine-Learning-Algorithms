## Scikit-learn Logistic Regression – Hyperparameters and Tuning Guide

### 1. penalty
--- 
 `Type: {'l1', 'l2', 'elasticnet', 'none'}`
   - Default: 'l2'
   - Purpose: Specifies the norm used in the penalization (regularization term).
   - Usage:
       1. 'l1' → Lasso (sparse weights, feature selection)
       2. 'l2' → Ridge (default, stable generalization)
       3. 'elasticnet' → combination of L1 and L2
       4. 'none' → no regularization
     
**Note**: Not all solvers support all penalties (see solver compatibility below).

### 2. C (inverse of regularization strength)
--- 
 - Type: float, default = 1.0
 - Purpose: Controls the `trade-off` between regularization and accuracy.
 - Interpretation:
    - `Small C` → strong regularization (simpler model)
    - `Large C` → weak regularization (risk of overfitting)

 - Tuning range: Try `0.001, 0.01, 0.1, 1, 10, 100` on a log scale

### 3. solver
--- 

- `Type: {'lbfgs', 'liblinear', 'sag', 'saga', 'newton-cg'}`
- Default: 'lbfgs'
- Purpose: Algorithm to optimize the cost function.
- Best choices:
   - `'liblinear'` → small datasets or L1 penalty
   - `'lbfgs', 'newton-cg'` → L2 penalty, multinomial loss, larger datasets
   - `'saga'` → supports L1, L2, and elasticnet; scalable

- Rule of thumb: use 'lbfgs' or 'saga' for large datasets

### 4. max_iter

--- 
- Type: int, default = 100
- Purpose: Maximum number of iterations for the solver to converge
- If you see a convergence warning, increase this to 200, 500, etc.

### 5. multi_class
--- 
- Type: {'auto', 'ovr', 'multinomial'}
- Default: 'auto'
- Purpose: Strategy for multiclass classification
   - 'ovr' (one-vs-rest): binary classifiers for each class
   - 'multinomial': softmax regression (requires solvers like 'lbfgs', 'saga')

### 6. class_weight
--- 

- Type: dict or 'balanced'
- Purpose: Adjusts weights inversely proportional to class frequencies
- 'balanced' → useful for imbalanced datasets

### 7. l1_ratio

--- 
- Type: float (between 0 and 1), used only with penalty='elasticnet'
- Purpose: Mix ratio for L1 and L2 (like α in elasticnet)
- l1_ratio=1.0 → pure L1
- l1_ratio=0.0 → pure L2

## LogisticRegression v/s LogisticRegressionCV 

| Feature               | `LogisticRegression`                                  | `LogisticRegressionCV`                                      |
| --------------------- | ----------------------------------------------------- | ----------------------------------------------------------- |
| Basic Function        | Fits logistic regression with fixed hyperparameters   | Performs logistic regression with built-in cross-validation |
| Hyperparameter Tuning | Manual – use `GridSearchCV` or `RandomizedSearchCV`   | Automatic – uses internal cross-validation to tune `C`      |
| Cross-Validation      | Not built-in                                          | Built-in via the `cv` parameter                             |
| Use Case              | Full control over all hyperparameters                 | Quick tuning of `C` with convenience                        |
| Tuning `C`            | Manually passed or tuned externally                   | Automatically selected from a list (e.g., `Cs=10`)          |
| Multiclass Strategy   | Supports `ovr` and `multinomial`                      | Supports `ovr` and `multinomial`                            |
| Output                | A single model                                        | Model with best `C` chosen from CV                          |
| Import Statement      | `from sklearn.linear_model import LogisticRegression` | `from sklearn.linear_model import LogisticRegressionCV`     |
