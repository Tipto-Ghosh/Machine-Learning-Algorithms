# Classification Metrics Explained

---

## 1. Confusion Matrix (Base for All)

Let's define our classes:
* **1 = Positive class** (e.g., person has disease)
* **0 = Negative class** (e.g., person is healthy)

Suppose you have these results on a test set of 10 samples:

| Actual / Predicted | Predicted: 1 | Predicted: 0 |
| :----------------- | :----------- | :----------- |
| **Actual: 1** | TP = 3       | FN = 2       |
| **Actual: 0** | FP = 1       | TN = 4       |

We use this table to calculate all other metrics.

---

## 2. Accuracy

$$Accuracy = \frac{TP + TN}{TP + TN + FP + FN}$$

**Using the above example:**

$$Accuracy = \frac{3 + 4}{3 + 4 + 1 + 2} = \frac{7}{10} = 0.70$$

**Tells:** Overall correctness of the model.

**Problem:** Can be misleading when classes are imbalanced.

---

## 3. Precision (Positive Predictive Value)

$$Precision = \frac{TP}{TP + FP}$$

**Using the above example:**

$$Precision = \frac{3}{3 + 1} = \frac{3}{4} = 0.75$$

**Tells:** Out of all instances predicted as positive, how many were actually correct.

**Use when false positives are costly:** For example, in cancer diagnosis, you don't want to wrongly tell someone they have cancer.

---

## 4. Recall (Sensitivity or True Positive Rate)

$$Recall = \frac{TP}{TP + FN}$$

**Using the above example:**

$$Recall = \frac{3}{3 + 2} = \frac{3}{5} = 0.60$$

**Tells:** Out of all actual positive instances, how many did the model correctly identify.

**Use when false negatives are costly:** For example, in detecting diseases, you don't want to miss sick patients.

---

## 5. F1-Score

$$F1 = 2 \cdot \frac{Precision \cdot Recall}{Precision + Recall}$$

**Using the above example:**

$$F1 = 2 \cdot \frac{0.75 \cdot 0.60}{0.75 + 0.60} = \frac{0.90}{1.35} \approx 0.6667$$

**Tells:** A balanced metric that considers both precision and recall.

**Best when you need a trade-off:** For example, in spam detection, you want to minimize both false positives (legitimate emails marked as spam) and false negatives (spam emails not caught).

---

## 6. Specificity (True Negative Rate)

$$Specificity = \frac{TN}{TN + FP}$$

**Using the above example:**

$$Specificity = \frac{4}{4 + 1} = \frac{4}{5} = 0.80$$

**Tells:** How well the model correctly identifies negative instances.

**Useful in medical tests:** For example, avoiding false alarms when a patient is healthy.

---

## 7. False Positive Rate (FPR)

$$FPR = \frac{FP}{FP + TN} = 1 - Specificity$$

**Using the above example:**

$$FPR = \frac{1}{1 + 4} = 0.20$$

**Tells:** How often the model wrongly predicts a positive (1) when the actual class is negative (0).

---

## 8. ROC Curve & AUC

* **ROC = Receiver Operating Characteristic:** Plots the True Positive Rate (Recall) against the False Positive Rate at various threshold settings.
    * **X-axis:** FPR
    * **Y-axis:** TPR (Recall)

* **AUC = Area Under Curve:** The area under the ROC curve.
    * Ranges from 0 to 1.
    * **1.0:** Represents a perfect model.
    * **0.5:** Represents a model performing no better than random guessing.

**Use AUC when you want to:**
    * Compare different models.
    * Evaluate overall model performance across all possible classification thresholds.

---

## Summary Table

| Metric      | Formula                                  | Use Case                                          |
| :---------- | :--------------------------------------- | :------------------------------------------------ |
| Accuracy    | $$\frac{TP + TN}{Total}$$                | Balanced data, overall correctness                |
| Precision   | $$\frac{TP}{TP + FP}$$                   | When false positives are costly                   |
| Recall      | $$\frac{TP}{TP + FN}$$                   | When false negatives are costly                   |
| F1 Score    | Harmonic mean of precision & recall      | Balanced metric, trade-off between precision & recall |
| Specificity | $$\frac{TN}{TN + FP}$$                   | Detecting negatives correctly                     |
| FPR         | $$\frac{FP}{FP + TN}$$                   | Used in ROC, measuring false alarms               |
| AUC-ROC     | Area under ROC curve                     | Overall model performance at all thresholds       |
