```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing(as_frame = True) 
X = housing.data
y = housing.target
```

```python
type(X)
```

```python
X.shape
```

```python
X = X.iloc[ : 1000 , : ]
```

```python
type(y)
```

```python
y = y[ : 1000]
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.model_selection import cross_val_score
```

```python
lr = LinearRegression() 
dt = DecisionTreeRegressor() 
svr = SVR()
```

```python
estimators = [('LR' , lr) , ('DT' , dt) , ('SVR' , svr)]
```

```python
# using single model 
for estimator in estimators: 
    scores = cross_val_score(estimator[1] , X , y , cv = 10 , scoring = 'r2')
    print(estimator[0] , np.mean(scores))
```

```python
from sklearn.ensemble import VotingRegressor

vr = VotingRegressor(estimators = estimators)
```

```python
np.mean(cross_val_score(vr , X , y , cv = 10 , scoring = 'r2'))
```

```python
for i in range(1 , 4): 
    for j in range(1 , 4): 
        for k in range(1 , 4): 
            weights = [i , j , k] 
            print(f"Weights: {weights}")
            score = np.mean(cross_val_score(VotingRegressor(estimators = estimators , weights = weights) , X , y , cv = 10)) 
            print(f"r2_score: {score}") 
            print('=' * 30)
```
