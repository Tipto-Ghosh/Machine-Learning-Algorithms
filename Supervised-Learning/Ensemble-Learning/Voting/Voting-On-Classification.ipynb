```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df = df[df['species'] != 0][['sepal_length' , 'sepal_width' , 'species']]
```

```python
df.head()
```

```python
df.shape
```

```python
X = df.drop(columns = ['species'])
y = df['species']
```

```python
X.shape
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import VotingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
```

```python
lr = LogisticRegression()
rf = RandomForestClassifier()
knn = KNeighborsClassifier()
```

```python
estimators = [
    ("LogisticRegression" , lr), 
    ("RandomForestClassifier" , rf), 
    ("KNeighborsClassifier" , knn) 
]
```

```python
# Find the accuracy of individual models
for estimator in estimators: 
    cross_val = np.mean(cross_val_score(estimator[1] , X , y , cv = 10 , scoring = 'accuracy'))
    print(f"Accuracy of {estimator[0]} is: {cross_val}")
```

### Hard Voting 

```python
vc = VotingClassifier(estimators = estimators)
cross_val = cross_val_score(vc , X , y , cv = 10 , scoring = 'accuracy')
print(np.mean(cross_val))
```

### Soft voting

```python
vc = VotingClassifier(estimators = estimators , voting = 'soft')
cross_val = cross_val_score(vc , X , y , cv = 10 , scoring = 'accuracy')
print(np.mean(cross_val))
```

### Weigthed Voting 

```python
for i in range(1 , 4): 
    for j in range(1 , 4): 
        for k in range(1 , 4): 
            weights = [i , j , k] # weights for our 3 models

            # with hard voting 
            vc1 = VotingClassifier(estimators = estimators , voting = 'hard' , weights = weights)
            cross_val1 = cross_val_score(vc1 , X , y , cv = 10 , scoring = 'accuracy')
            cv1 = np.mean(cross_val1) 
            # print(' - ' * 30)
            print("Hard Voting")
            print(f"Weights: {weights} mean cross val: {cv1}")
            print(' - ' * 30)

            # with soft voting 
            vc2 = VotingClassifier(estimators = estimators , voting = 'soft' , weights = weights)
            cross_val2 = cross_val_score(vc2 , X , y , cv = 10 , scoring = 'accuracy')
            cv2 = np.mean(cross_val2) 
            # print(' - ' * 30)
            print("Soft Voting")
            print(f"Weights: {weights} mean cross val: {cv2}")
            print('-' * 30)
```
