```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

### Load the dataset

```python
from sklearn.datasets import load_diabetes
X , y = load_diabetes(return_X_y = True)
```

```python
X.shape
```

```python
# Do the train test split 
from sklearn.model_selection import train_test_split

X_train , X_test , y_train , y_test = train_test_split(X , y , test_size = 0.2 , random_state = 22)
```

## Use sklearn Ridge Regression

```python
from sklearn.linear_model import Ridge

sklearn_ridge = Ridge(alpha = 0.1 , solver = 'cholesky')
```

```python
sklearn_ridge.fit(X_train , y_train)
```

```python
y_pred_sklearn = sklearn_ridge.predict(X_test)
```

```python
from sklearn.metrics import r2_score

r2_score_sk_ridge = r2_score(y_test , y_pred_sklearn)

print("r2 score: " , r2_score_sk_ridge)
```

```python
print("Coef_: " , sklearn_ridge.coef_)
print("intercept_ : " , sklearn_ridge.intercept_)
```

## let's impletement out own Ridge Class

```python
class RidgeRegression: 
    def __init__(self , alpha = 0.1): 
        self.alpha = alpha #  strength of ridge regression
        self.coef_ = None 
        self.intercept_ = None 
    def fit(self , X_train , y_train):
        X_train = np.asarray(X_train)
        y_train = np.asarray(y_train)
        
        # Add a column of one's in-front of X_train (Bias Term)
        X_bias = np.hstack([np.ones((X_train.shape[0] , 1)) , X_train])

        # Identity matrix(excluse the bias from regularization)
        I = np.eye(X_bias.shape[1])
        I[0 , 0] = 0 # Dont need to regularize the bias term 
        
        # Calculate the equation 
        A = X_bias.T @ X_bias + self.alpha * I 
        B = X_bias.T @ y_train

        weights = np.linalg.inv(A) @ B 
        
        # set coef_ and intercept_
        self.intercept_ = weights[0]
        self.coef_ = weights[1 : ]

        print("Coef_: " , sklearn_ridge.coef_)
        print("\n --------------------------------------\n")
        print("intercept_ : " , sklearn_ridge.intercept_)
        
    def predict(self , X_test): 
        X_test = np.asarray(X_test)
        return np.dot(X_test , self.coef_) + self.intercept_
```

```python
# Train the model 

myRidge = RidgeRegression()

myRidge.fit(X_train , y_train)
```

```python
y_pred = myRidge.predict(X_test)
```

```python
print("R2 Score: " , r2_score(y_test , y_pred))
```
