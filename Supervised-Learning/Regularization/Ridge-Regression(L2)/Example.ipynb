## When to Choose Ridge Regression

Use Ridge Regression when:

 - `Multicollinearity exists`:
    - If features are highly correlated, ordinary least squares (OLS) can produce unstable coefficients.
    - Ridge adds a penalty to reduce model complexity and stabilize the solution.

 - `Want to prevent overfitting`:
    - When there are more features than samples or many small/irrelevant features, Ridge helps reduce model variance by shrinking the
      coefficients.

 - `Want to keep all features`:
    - Unlike Lasso, which can set coefficients to zero (feature elimination), Ridge shrinks coefficients but retains all features.
 - Suspect model is too sensitive to noise:
   - Regularization (like Ridge) helps generalize better on unseen data.

 - All Predictors Potentially Relevant: Unlike Lasso Regression, Ridge tends to shrink coefficients but not completely eliminate them. This
   suggests that all features contribute to some extent.

--- 
$$ \text{Cost Function} = \text{RSS} + \alpha \sum_{j=1}^{p} w_j^2 $$
--- 

## Why it's called 'Ridge'
The name 'ridge' comes from two main points:

- `Mathematical Similarity`: It's inspired by `ridge analysis`, a statistical technique for visualizing response surface equations.
- Adding a `Ridge` to the Matrix: Mathematically, Ridge Regression adds a small positive constant (α or λ) to the main diagonal of the X.T.X
  X matrix (where X is your design matrix). This `ridge` makes the matrix invertible, even if X.T.X itself is singular due to
  multicollinearity, thus stabilizing the solution.

---

## Why “L2 Regularization”?
Ridge adds a penalty term to the loss function proportional to the square of the magnitude of coefficients.

 
This is called L2 norm: the sum of squares of coefficients.

It tries to minimize both the error and the size of the coefficients.

## How to Choose the Best Value of α (Alpha)

- What is alpha?
  - alpha (or λ) controls the strength of the penalty:
  - High alpha → more shrinkage → lower variance but possibly underfitting.
  - Low alpha → less shrinkage → closer to linear regression → risk of overfitting.
- Cross-Validation: The most effective method is cross-validation (e.g., K-fold). You'll train and evaluate your model with a range of α values, selecting the one that gives the best performance on the validation sets (e.g., the lowest mean squared error).
