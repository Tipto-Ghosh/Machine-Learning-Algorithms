```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

#### Load the Data

```python
from sklearn.datasets import load_diabetes
data = load_diabetes()
df = pd.DataFrame(data.data , columns = data.feature_names)
df['target'] = data.target
```

```python
df.head()
```

```python
df.shape
```

```python
X = df.drop(columns = ['target'] , axis = 1)
y = df['target']
```

```python
from sklearn.model_selection import train_test_split

X_train , X_test , y_train , y_test = train_test_split(X , y , test_size = 0.2 , random_state = 2)
```

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score
```

```python
coefs = []
r2_scores = []

for alpha_value in [0 , 10 , 100 , 1000]:
    reg = Ridge(alpha = alpha_value)
    reg.fit(X_train , y_train)
    
    y_pred = reg.predict(X_test)
    r2_scores.append(r2_score(y_test , y_pred))

    coefs.append(reg.coef_.tolist())
```

```python
r2_scores
```

<h2 style = "color:red;">How the coefficients get affected by alpha?</h2>

```python
plt.figure(figsize = (14 , 9))

plt.subplot(221)
plt.bar(data.feature_names , coefs[0])
plt.title(f'Aplha = 0 , r2_score = {round(r2_scores[0] , 2)}')

plt.subplot(222)
plt.bar(data.feature_names , coefs[1])
plt.title(f'Aplha = 10 , r2_score = {round(r2_scores[1] , 2)}') 

plt.subplot(223)
plt.bar(data.feature_names , coefs[2])
plt.title(f'Aplha = 100 , r2_score = {round(r2_scores[2] , 2)}') 

plt.subplot(224)
plt.bar(data.feature_names , coefs[3])
plt.title(f'Aplha = 1000 , r2_score = {round(r2_scores[3] , 2)}') 

plt.show()
```

<h3 style = "color:red;">Note: </h3>


- From the graphs for all different values of alpha we can see that as value of alpha is increasing the value of coeffiecents goes towards to zero but never been zero. 

<h2 style = "color:green;">Higher Values of Coeffiecents are impacted more!!</h2>

<h2 style = "color:green;">Bias Variance Tradeoff</h2>

- As value of alpha increases -> Bias Decrese and model became overfitted and Variance Increases.
- As value of alpha decreses -> Bias Increase and model became underfitted.


```python
m = 100
X = 5 * np.random.rand(m , 1) - 2
y = 0.7 * X**2 + 3 + np.random.rand(m , 1)

plt.scatter(X , y) 
plt.show()
```

```python
X_train , X_test , y_train , y_test = train_test_split(X.reshape(100 , 1) , y.reshape(100) , test_size = 0.2 , random_state = 23)
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly = PolynomialFeatures(degree = 15)

X_train = poly.fit_transform(X_train)
X_test = poly.transform(X_test)
```

```python
from mlxtend.evaluate import bias_variance_decomp
```

```python
alphas = np.linspace(0 , 30 , 100)

loss = []
bias = []
variance = []

for i in alphas: 
    reg = Ridge(alpha = i) 

    avg_expected_loss , avg_bias , avg_var = bias_variance_decomp(
        reg , X_train , y_train , X_test , y_test , loss = 'mse' , random_seed = 123
    )

    loss.append(avg_expected_loss)
    bias.append(avg_bias)
    variance.append(avg_var)
```

```python
plt.plot(alphas , loss , label = 'loss')
plt.plot(alphas , bias , label = 'bias') 
plt.plot(alphas , variance , label = 'variance') 

# plt.ylim(0 , 5)
plt.xlabel('alpha')
plt.legend()
plt.show()
```
