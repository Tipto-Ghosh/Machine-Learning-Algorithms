```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score , classification_report
from sklearn.tree import DecisionTreeClassifier
```

```python
from sklearn.datasets import make_classification

X , y = make_classification(n_features = 5 , n_redundant = 0 , n_informative = 5 , n_clusters_per_class = 1)
```

```python
df = pd.DataFrame(X , columns = ['col1' , 'col2' , 'col3' , 'col4' , 'col5'])
df['target'] = y 
df.shape
```

```python
df.head()
```

```python
# function for row sampling 
def sample_rows(df , percent , replace = True): 
    return df.sample(int(percent * df.shape[0]) , replace = replace)
```

```python
# function for feature sampling 
def sample_cols(df , percent , replace = True): 
    col_list = df.columns.tolist()[ : -1]
    sample_col_count = int(percent * (df.shape[1] - 1))
    cols = random.sample(col_list , sample_col_count)
    return df[cols]
```

```python
# function for both row and feature sampling 
def combined_sampling(df , row_percent , col_percent , replace = True): 
    new_df = sample_cols(df , col_percent , replace)
    return sample_rows(new_df , row_percent , replace)
```

```python
# In bagging we train same type of model with different subset of data 

# make 3 datasets for 3 models
df1 = sample_rows(df , 0.1)
df2 = sample_rows(df , 0.1) 
df3 = sample_rows(df , 0.1)
```

```python
dt1 = DecisionTreeClassifier() 
dt2 = DecisionTreeClassifier() 
dt3 = DecisionTreeClassifier()
```

```python
dt1.fit(df1.iloc[ : , : 5] , df1['target'])
dt2.fit(df2.iloc[ : , : 5] , df2['target']) 
dt3.fit(df3.iloc[ : , : 5] , df3['target']) 
```
