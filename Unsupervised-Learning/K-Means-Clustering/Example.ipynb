```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
from matplotlib import pyplot as plt 
```

```python
df = pd.read_csv("student_clustering.csv")
df.shape
```

```python
df.head()
```

```python
plt.scatter(df['cgpa'] , df['iq'] , marker = 'o')
plt.xlabel("cgpa")
plt.ylabel("iq")
plt.title("cgpa v/s iq")
plt.show()
```

```python
from sklearn.cluster import KMeans
```

```python
wss = []

for i in range(1 , 11): 
    km = KMeans(n_clusters = i) 
    km.fit_predict(df) 
    wss.append(km.inertia_)
```

```python
wss
```

```python
k = range(1 , 11)
plt.plot(k , wss) 
plt.title("K v/s WSS")
plt.xlabel("number of clusters")
plt.ylabel("WSS")
plt.show()
```

```python
X = df.iloc[ : , : ].values
```

```python
km = KMeans(n_clusters = 4)
y_means = km.fit_predict(X)
```

```python
y_means
```

```python
plt.scatter(X[y_means == 0 , 0] , X[y_means == 0 , 1] , c = 'blue')
plt.scatter(X[y_means == 1 , 0] , X[y_means == 1 , 1] , c = 'red') 
plt.scatter(X[y_means == 2 , 0] , X[y_means == 2 , 1] , c = 'green') 
plt.scatter(X[y_means == 3 , 0] , X[y_means == 3 , 1] , c = 'orange')
plt.show()
```
