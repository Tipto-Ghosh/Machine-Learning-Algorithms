```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
from matplotlib import pyplot as plt 
import warnings
warnings.filterwarnings("ignore")
pd.set_option('display.max_columns' , None)
pd.options.display.float_format = '{:20.2f}'.format
```

```python
customer_data = pd.read_csv('hierarchical-clustering-with-python-and-scikit-learn-shopping-data.csv')
```

```python
customer_data.head()
```

```python
customer_data.shape
```

```python
# taking only last 2 columns 
data = customer_data.iloc[ : , 3 : 5].values
data.shape
```

```python
# make the dendogram
import scipy.cluster.hierarchy as shc

plt.figure(figsize = (10 , 7))
plt.title("Customer Dendograms")

dend = shc.dendrogram(shc.linkage(data , method = 'ward'))
```

```python
plt.figure(figsize = (10 , 7))
plt.title("Customer Dendograms")

dend = shc.dendrogram(shc.linkage(data , method = 'single'))
```

```python
plt.figure(figsize = (10 , 7))
plt.title("Customer Dendograms")

dend = shc.dendrogram(shc.linkage(data , method = 'complete'))
```

```python
plt.figure(figsize = (10 , 7))
plt.title("Customer Dendograms")

dend = shc.dendrogram(shc.linkage(data , method = 'average'))
```

```python
from sklearn.cluster import AgglomerativeClustering

cluster = AgglomerativeClustering(
    n_clusters = 5 , linkage = 'ward' , metric = 'euclidean'
)
labels = cluster.fit_predict(data)
```

```python
labels
```

```python
plt.figure(figsize = (10 , 6))
plt.scatter(x = data[ : , 0] , y = data[ : , 1] , c = cluster.labels_ , cmap = "viridis")
plt.show()
```
